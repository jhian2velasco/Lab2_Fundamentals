```python
#Personal Identification
SURNAME = "VELASCO"
SEED_NUM = 12345

# Generate a unique password
password_number = (sum(ord(char) for char in SURNAME) + SEED_NUM) % 10000

generated_password = SURNAME[:3].upper() + str(password_number)

#Generate an attempt limit
attempt_limit = (SEED_NUM % 4) + 3

print("Authentication System")
print ("--------------------")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)

attempts = 0
access_granted = False

while attempts < attempt_limit:

    entered_password = input("\nEnter password: ")
    attempts += 1
    if entered_password == generated_password:
        access_granted = True
        print ("Access Granted!")
        break
    else:
        remaining = attempt_limit - attempts
        
        if remaining > 0:
            print("Incorrect password")
            print("Attempts remaining:", remaining)
        else:
            print("Incorrect password")
            print("Maximum attempts reached.")

# Determine final system state
if access_granted:
    system_state = "UNLOCKED"
    access_results = "ACCESS GRANTED"
else:
    system_state = "LOCKED"
    access_results = "ACCESS DENIED"

#Final Output
print("\nAuthentication Summary")
print("----------------------")
print("Generated Password:", attempt_limit)
print("Attempts Made:", attempts)
print("Access Results:", access_results)
print("Final System State:", system_state)
```
