```python
import random

#Personal Identification
SURNAME = "Velasco"
SEED_NUM = 12345

#Use surname and seed to create a unique random sequence
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

#Generate 10 sensor readings
sensor_data = []

for i in range(10):
    #Some readings are intentionally invalid
    if i == 3:
        sensor_data.append("invalid")
    else:
        sensor_data.append(round(random.uniform(0, 100), 2))

print("Generated Sensor Data")
print(sensor_data)

#Process and classify readings
valid_results = []
invalid_results = []
classification_results = []

for reading in sensor_data:
    try:
        value = float(reading)
        # Example Operating Ranges
        if 0 <= value <= 30:
            classification = "LOW"
        elif 31 <= value <= 70:
            classification = "NORMAL"
        elif 71 <= value <= 100:
            classification = "HIGH"
        else:
            classification = "OUT OF RANGE"

        valid_results.append(value)
        classification_results.append((value, classification))
    except ValueError:
        invalid_results.append(reading)

# Display results
print("\nValid Readings:")
print(valid_results)

print("\nInvalid Readings:")
print(invalid_results)

print("\nClassification Results:")
for value, classification in classification_results:
    print(f"{value} -> {classification}")

print("\nSummary:")
print("Total readings:", len(sensor_data))
print("Valid readings:", len(valid_results))
print("Invalid readings:", len(invalid_results))
```
