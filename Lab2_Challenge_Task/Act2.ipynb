```python
import random
import string

#Personal Identification
SURNAME = "VELASCO"
SEED_NUM = 12345

# Create a unique seed
seed_value = sum(ord(char) for char in SURNAME) + SEED_NUM
random.seed(seed_value)

#Generate a unique signal
characters = string.ascii_letters + string.digits + " @#$%!"

signal = ""
for i in range(20):
    signal += random.choice(characters)
print("Generated Signal:")
print(signal)

# Process and analyze the signal
letters = 0
numbers = 0
spaces = 0
special_characters = 0

print("\nCharacter Analysis:")
for char in signal:
    if char.isalpha():
        character_type = "Letter"
        letters += 1
    elif char.isdigit():
        character_type = "Number"
        numbers += 1

    elif char.isspace():
        character_type = "Space"
        spaces += 1
    else:
        character_type = "Special Character"
        special_characters += 1

    print(f"{char} -> {character_type}")

# Display Final Analysis
print("\nDiagnostic Report")
print("--------------")
print("Letters:", letters)
print("Numbers:", numbers)
print("Spaces:", spaces)
print("Special Characters:", special_characters)

#Simple classification
if special_characters > 0:
    signal_classification = "MIXED / SPECIAL CHARACTER SIGNAL"
elif letters > 0:
    signal_classification = "ALPHANUMERIC SIGNAL"
elif numbers > 0:
    signal_classification = "LETTER SIGNAL"
else:
    signal_classification = "OTHER SIGNAL"
print("Signal Classification:", signal_classification)
```
